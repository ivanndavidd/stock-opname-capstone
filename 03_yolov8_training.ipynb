## 1. Install YOLOv8 (Ultralytics)

```python
# Import libraries
from ultralytics import YOLO
import torch
import os
import yaml
from pathlib import Path

print(f"✓ Ultralytics version: {torch.__version__}")
print(f"✓ CUDA available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"✓ GPU: {torch.cuda.get_device_name(0)}")
else:
    print("⚠️  Using CPU (training akan lambat)")
```

## 2. Configuration

```python
# Paths
DATA_YAML = "./data/data.yaml"  # Changed: No SKU110K subfolder
WEIGHTS_DIR = "./weights"
RUNS_DIR = "./runs"

# Training parameters
MODEL_SIZE = 'yolov8m.pt'  # Options: yolov8n, yolov8s, yolov8m, yolov8l, yolov8x
EPOCHS = 100
BATCH_SIZE = 16  # Adjust based on GPU memory
IMG_SIZE = 640
PATIENCE = 50  # Early stopping patience

# Create directories
os.makedirs(WEIGHTS_DIR, exist_ok=True)
os.makedirs(RUNS_DIR, exist_ok=True)

print("✓ Configuration set:")
print(f"  • Model: {MODEL_SIZE}")
print(f"  • Epochs: {EPOCHS}")
print(f"  • Batch size: {BATCH_SIZE}")
print(f"  • Image size: {IMG_SIZE}")
print(f"  • Data config: {DATA_YAML}")
```

## 3. Verify Data Configuration

```python
# Load and verify data.yaml
with open(DATA_YAML, 'r') as f:
    data_config = yaml.safe_load(f)

print("📄 Data configuration:")
for key, value in data_config.items():
    print(f"  {key}: {value}")

# Verify paths exist
base_path = Path(data_config['path'])
train_path = base_path / data_config['train']
val_path = base_path / data_config['val']

print(f"\n✓ Verifying paths:")
print(f"  Train: {train_path} {'✓' if train_path.exists() else '✗'}")
print(f"  Val: {val_path} {'✓' if val_path.exists() else '✗'}")
```

## 4. Load Pretrained Model

```python
# Load pretrained YOLO model
print(f"📦 Loading {MODEL_SIZE}...")
model = YOLO(MODEL_SIZE)

print(f"✓ Model loaded successfully!")
print(f"\n📊 Model info:")
model.info()
```

## 5. Training Configuration Details

```python
# Training arguments
train_args = {
    'data': DATA_YAML,
    'epochs': EPOCHS,
    'batch': BATCH_SIZE,
    'imgsz': IMG_SIZE,
    'patience': PATIENCE,
    'save': True,
    'save_period': 10,  # Save checkpoint every 10 epochs
    'cache': False,  # Set True if enough RAM
    'device': 0 if torch.cuda.is_available() else 'cpu',
    'workers': 8,
    'project': RUNS_DIR,
    'name': 'sku110k_yolov8m',
    'exist_ok': True,
    'pretrained': True,
    'optimizer': 'SGD',
    'verbose': True,
    'seed': 42,
    'deterministic': True,
    'single_cls': True,  # Single class detection
    'rect': False,  # Rectangular training
    'cos_lr': True,  # Cosine learning rate scheduler
    'close_mosaic': 10,  # Disable mosaic augmentation in last 10 epochs
    'resume': False,  # Resume from last checkpoint
    'amp': True,  # Automatic Mixed Precision
    'fraction': 1.0,  # Use 100% of data (reduce for quick test)
    'profile': False,  # Profile model
}

print("⚙️ Training configuration:")
for key, value in train_args.items():
    print(f"  {key}: {value}")
```

## 6. Quick Test Training (Optional)

Test dengan subset kecil dulu untuk verify setup

```python
# Optional: Quick test dengan 3 epochs dan 10% data
RUN_TEST = False  # Set True untuk test

if RUN_TEST:
    print("🧪 Running quick test...")
    test_args = train_args.copy()
    test_args['epochs'] = 3
    test_args['fraction'] = 0.1
    test_args['name'] = 'sku110k_test'
    
    test_model = YOLO(MODEL_SIZE)
    test_results = test_model.train(**test_args)
    
    print("✓ Test training complete!")
    print("If successful, proceed to full training.")
else:
    print("⏭️  Skipping test training")
```

## 7. Start Full Training 🚀

```python
print("="*70)
print("🚀 STARTING FULL TRAINING")
print("="*70)
print(f"\nThis will take approximately 12-18 hours with GPU")
print(f"Monitor progress in: {RUNS_DIR}/sku110k_yolov8m/")
print(f"\nPress Ctrl+C to stop (model will save checkpoint)\n")

# Start training
results = model.train(**train_args)

print("\n" + "="*70)
print("✅ TRAINING COMPLETE!")
print("="*70)
```

## 8. Training Results Analysis

```python
# Load best model
best_model_path = f"{RUNS_DIR}/sku110k_yolov8m/weights/best.pt"
best_model = YOLO(best_model_path)

print(f"✓ Best model loaded from: {best_model_path}")
```

```python
# Display training plots
from IPython.display import Image, display
import glob

print("📊 Training Results:")

# Find and display plots
results_dir = f"{RUNS_DIR}/sku110k_yolov8m"
plot_files = [
    'results.png',
    'confusion_matrix.png',
    'P_curve.png',
    'R_curve.png',
    'PR_curve.png',
    'F1_curve.png'
]

for plot_file in plot_files:
    plot_path = os.path.join(results_dir, plot_file)
    if os.path.exists(plot_path):
        print(f"\n{plot_file}:")
        display(Image(filename=plot_path))
    else:
        print(f"⚠️  {plot_file} not found")
```

## 9. Validation on Val Set

```python
# Validate model
print("🔍 Validating model on validation set...")

val_results = best_model.val(
    data=DATA_YAML,
    split='val',
    batch=BATCH_SIZE,
    imgsz=IMG_SIZE,
    device=0 if torch.cuda.is_available() else 'cpu',
    save_json=True,
    save_hybrid=True,
    conf=0.25,
    iou=0.6,
    max_det=300,  # Max detections per image
    plots=True
)

print("\n✓ Validation complete!")
```

```python
# Print metrics
print("\n📊 Validation Metrics:")
print(f"  • mAP@0.5: {val_results.box.map50:.4f}")
print(f"  • mAP@0.5:0.95: {val_results.box.map:.4f}")
print(f"  • Precision: {val_results.box.mp:.4f}")
print(f"  • Recall: {val_results.box.mr:.4f}")
print(f"  • F1-Score: {2 * (val_results.box.mp * val_results.box.mr) / (val_results.box.mp + val_results.box.mr):.4f}")
```

## 10. Test Inference on Sample Images

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

def visualize_predictions(image_path, model, conf=0.25):
    """
    Run inference and visualize results
    """
    # Run inference
    results = model.predict(
        source=image_path,
        conf=conf,
        iou=0.6,
        max_det=300,
        save=False,
        device=0 if torch.cuda.is_available() else 'cpu'
    )
    
    # Get annotated image
    annotated = results[0].plot()
    
    # Convert BGR to RGB
    annotated_rgb = cv2.cvtColor(annotated, cv2.COLOR_BGR2RGB)
    
    # Display
    plt.figure(figsize=(15, 10))
    plt.imshow(annotated_rgb)
    plt.axis('off')
    plt.title(f'Detections: {len(results[0].boxes)} objects (conf > {conf})', fontsize=14)
    plt.tight_layout()
    plt.show()
    
    return results

print("✓ Visualization function ready")
```

```python
# Test on random validation images
val_images_dir = "./data/images/val"  # Changed: No SKU110K subfolder
sample_images = [os.path.join(val_images_dir, f) for f in os.listdir(val_images_dir) if f.endswith('.jpg')][:5]

print(f"🔍 Testing on {len(sample_images)} sample images...\n")

for img_path in sample_images[:3]:  # Show 3 samples
    print(f"Image: {os.path.basename(img_path)}")
    results = visualize_predictions(img_path, best_model, conf=0.25)
    print()
```

## 11. Export Model

```python
# Export model to different formats
print("📦 Exporting model...")

# Export to ONNX (for deployment)
onnx_path = best_model.export(format='onnx')
print(f"✓ ONNX model: {onnx_path}")

# Export to TorchScript (for PyTorch deployment)
# torchscript_path = best_model.export(format='torchscript')
# print(f"✓ TorchScript model: {torchscript_path}")

# Copy best.pt to weights folder
import shutil
final_weights = os.path.join(WEIGHTS_DIR, 'sku110k_best.pt')
shutil.copy(best_model_path, final_weights)
print(f"✓ Weights copied to: {final_weights}")
```
